в случае вызова от класса это функция, во втором случае - метод экземпляра.
Метод отличается от функции:
- метод - функция, которая объявлена внутри класса
- метод привязан к конкретному объекту, тогда как функция не связана. То есть метод `speak()` нельзя вызвать просто так, только от объекта 
- при вызове метода автоматически объект будет подставляться в качетсве аргумента метода. Собсвтенно об этом и ошбика выше. В нашем случае объект `barsik` - экземпляр, подставляется в качестве аргумента


```python
class Cat:
    def speak(*args):
        print('meow meow')
        print(args)
```

```python
barsik = Cat()
barsik.speak()
```

```python
# адреса в памяти одинаковые
print(f'адреса экземпляра {barsik}')
barsik.speak()
```

Что можно заметить:
- ссылка на аргумент метода совпадает с адресом экземпляра класса, то есть экземпляр подставляется в качетсве аргумента
- следовательно `speak` это метод экземпляра класса, а не атрибут-функция

Создайте класс Point. У него должны быть реализованы следующие методы:
- Метод set_coordinates, который принимает координаты точки на плоскости и сохраняет их в атрибуты экземпляра x и y;
- Метод get_distance_to_origin, который возвращает расстояние между точкой и началом координат
- Метод display, который:
- - выводит сообщение «Координаты не заданы», если у точки отсутствуют атрибуты x и y;
- - выводит сообщение «Point(x, y)», если атрибуты x и y заданы. Вместо x и y нужно подставить соответствующие значения координат.

```python
class Point:
    def set_coordinates(self, x, y):
        self.x = x
        self.y = y

    def get_distance_to_origin(self):
        if hasattr(self, 'x') and hasattr(self, 'y'):
            return (self.x ** 2 + self.y ** 2) ** 0.5
        else:
            return None
        
    def display(self):
        if hasattr(self, 'x') and hasattr(self, 'y'):
            print(f'Point({self.x}, {self.y})')
        else:
            print('Координаты не заданы')
```

```python
p1 = Point()
p1.set_coordinates(6, 8)
p1.display()
print(p1.get_distance_to_origin())
```

```python
p2 = Point()
p2.display()
p2.set_coordinates(3, 4)
p2.display()
print(p2.get_distance_to_origin())
```

```python
p3 = Point()
p3.display()
print(p3.get_distance_to_origin())
p3.x = 4
p3.display()
print(p3.get_distance_to_origin())
p3.y = 3
p3.display()
print(p3.get_distance_to_origin())
```

# Инициализация экземпляра
В примере выше для определения атрибутов в экземпляре создалавалась функция set_coordinates, куда передавались атрибуты x и y. Однако при создании атрибута класса можно сделать это сразу, передавая атрибуты в качестве аргументов. Для этого сущесвует магический метод `__init__` (initialization).

Особенность магических методов в том, что они вызываются в определенный момент (по определенному событию - оно обычно закреплено). Метод `__init__` вызывается при создании объекта. Рассмотрим пример:

```python
class Car:
    def __init__(self):
        print(f'привет, новый атрибут {self}')

car_0 = Car()
```

Как видно выше, функция отработала сразу при инициализации. В качестве аргументов в метод передаются все необходимые атрибуты экземпляра, при чем они передаются как аргументы класса - по позиции или по имени:

```python
class Car:
    def __init__(self, new_model, new_engine, new_horse_power, new_color):
        self.model = new_model
        self.engine = new_engine
        self.horse_power = new_horse_power
        self.color = new_color

car_0 = Car('bmw', '2.0', 200, 'white')
```

```python
car_0.color, car_0.engine, car_0.horse_power, car_0.model
```

```python
car_0.__dict__
```

Так же можно указывать значения атрибутов по умолчанию:

```python
class Car:
    def __init__(self, new_model='audi', new_engine=2.0, new_horse_power=100, new_color='red'):
        self.model = new_model
        self.engine = new_engine
        self.horse_power = new_horse_power
        self.color = new_color

car_0 = Car()
car_0.__dict__
```

Обычно принято, что входящие параметры называются так же, как и атрибуты экземпляра, поэтому правильнее писать так:

```python
class Car:
    def __init__(self, model='audi', engine=2.0, horse_power=100, color='red'):
        self.model = model
        self.engine = engine
        self.horse_power = horse_power
        self.color = color

car_0 = Car()
car_0.__dict__
```

Рассмотрим так же случай, когда ранее мы объявляли атрибуты внутри класса и в чем разница значения атрибутов по умолчанию в `__init__` и самом классе

Как видно, `speed`, который меняется в классе, никак не меняется у существующего объекта, потому что он объявлен в методе `__init__`, то есть привязан к экземпляру класса. Атрибут `model` привязан к классу, и в случае смены значения по умолчанию, значения атрибута поменяется у всех экземпляров. Поэтому такой способ определения атрибутов класса используется для объявления констант, когда это значение одинаковые для всех экземпляров класса

# принцип DRY
Принцип DRY (Don’t repeat yourself) является принципом разработки, когда идет избегание повторной логики в коде. Далее пример с классом `Point`, с помощью которого можно будет создавать точки с координатами, перемещать их и тд



```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
```

```python
p1 = Point(3, 4)
p2 = Point(-1, -2)
```

Хотелось бы не всегда передавать в класс аргументы - сделаем по умолчанию 0 0

```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y
```

```python
p3 = Point()
```

Теперь зададим точкам поведение. Зададим перемещение:

```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def move(self, new_x, new_y):
        self.x = new_x
        self.y = new_y
```

```python
p4 = Point()
p4.move(1, 2)
p4.__dict__
```

Теперь добавим еще поведения - обнулим координаты точки

```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def move(self, new_x, new_y):
        self.x = new_x
        self.y = new_y

    def home(self):
        self.x = 0
        self.y = 0
```

Что общего у этих методов? - код практически одинаковый, логика так же одинаковая - задаем аргументы, которые становятся атрибутами точки. Если `__init__` создает атрибуты автоматически, то методы `home` и `move` различаются только тем, что в первом значения по умолчанию будут (0, 0), тогда как в move мы бы явно их задавали.

Это и есть дублирование логики. Чтобы ее избежать, в функциях класса можно вызывать дургие функции класса. Функция внутри класса может вызывать другую функцию класса как метод от объекта, то есть все как с атрибутами - нужен `self.`

```python
class Point:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def move(self, new_x, new_y):
        self.x = new_x
        self.y = new_y

    def home(self):
        self.move(0, 0)
```

Аналогично можно модифицировать метод `__init__`

```python
class Point:
    def __init__(self, x=0, y=0):
        self.move(x, y)

    def move(self, new_x, new_y):
        self.x = new_x
        self.y = new_y

    def home(self):
        self.move(0, 0)
```

```python
p5 = Point()
p5.__dict__
```

```python
p5.move(1, 2)
p5.__dict__
```

```python
p5.home()
p5.__dict__
```

Этот пример показывает принцип dry - основной функционал перешел к методу `move()`

Далее модицифицируем метод и сделаем так, чтобы работать с нескокими точками

```python
class Point:
    def __init__(self, x=0, y=0):
        self.move(x, y)

    def move(self, new_x, new_y):
        self.x = new_x
        self.y = new_y

    def home(self):
        self.move(0, 0)

    def print_coord(self):
        print(f'координаты точки ({self.x}, {self.y})')
```

```python
p6 = Point()
p6.move(1, 2)
p6.print_coord()
```

```python
class Point:
    def __init__(self, x=0, y=0):
        self.move(x, y)

    def move(self, new_x, new_y):
        self.x = new_x
        self.y = new_y

    def home(self):
        self.move(0, 0)

    def print_coord(self):
        print(f'координаты точки ({self.x}, {self.y})')

    def calc_distance(self, another_point):
        # нужно, чтобы объект another_point принадлежал классу Point
        if isinstance(another_point, Point):
            return ((another_point.x - self.x) ** 2 + (another_point.y - self.y) ** 2) ** 0.5
        else:
            raise ValueError("аргумент не принадлежит классу Point")
```

```python
p7 = Point()
p8 = Point(3, 4)

p7.calc_distance(p8)
```

```python
p8.calc_distance(p7)
```

```python
p8.calc_distance((0, 0))
```

Добавим к классу Point собственные атрибуты. Например при каждй инициализации точки мы хотим добавлять ее в список класса. Для этого в `__init__` надо обратиться к атрибуту класса, который будет распространятся на все экземпляры. Делается это через `класс.атрибут`

```python
class Point:
    list_points = []

    def __init__(self, x=0, y=0):
        self.move(x, y)
        Point.list_points.append(self) # добавляет экземпляры при инициализации

    def move(self, new_x, new_y):
        self.x = new_x
        self.y = new_y

    def home(self):
        self.move(0, 0)

    def print_coord(self):
        print(f'координаты точки ({self.x}, {self.y})')

    def calc_distance(self, another_point):
        # нужно, чтобы объект another_point принадлежал классу Point
        if isinstance(another_point, Point):
            return ((another_point.x - self.x) ** 2 + (another_point.y - self.y) ** 2) ** 0.5
        else:
            raise ValueError("аргумент не принадлежит классу Point")
```

```python
p9 = Point()
```

```python
# адреса совпадают
Point.list_points, p9
```

```python
p10 = Point(1, 2)
Point.list_points, p10
```
