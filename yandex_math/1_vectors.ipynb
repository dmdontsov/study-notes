# Нормы векторов

```python
from numpy import linalg
```

### L1 норма
сумма по модулю всех координат вектора
$\\|\vec{a}|\\_1 = |a_1| + |a_2| + \dots + |a_n| = \sum_{i=1}^n |a_i|$ — $L_1$ норма;

```python
a = np.array([1, 3, 5])

# второй параметр - порядковый номер нормы
print(linalg.norm(a, 1))
```

Задача: в таблице собраны прайс-листы на некоторые материалы, которые компания закупает у двух поставщиков.

|  | 1 кг гвоздей | литр краски | молотки  |  паркет |
|-----------|-----|-------|-------|-------|
| магазин_1    | 150  | 2300    | 500  |  5000 |
| магазин_2    | 200  | 2500    | 600  |  4000 |

Необходимо рассчитать L1 норму для каждого магазина

```python
shop_1 = np.array([100, 2300, 1000, 5000])
shop_2 = np.array([50, 2200, 600, 5700])
```

```python
print(f'L1 for shop_1: {linalg.norm(shop_1, 1)}')
print(f'L1 for shop_2: {linalg.norm(shop_2, 1)}')
```

### L2 норма
корень из суммы квадратов всех координат
$\\|\vec{a}|\\_2 = \sqrt{a_1^2 + a_2^2 + \dots + a_n^2} = \sqrt{\sum_{i=1}^n a_i^2}$ — $L_2$ норма;

```python
b = np.array([3, 4, 10])

# второй параметр - порядковый номер нормы
print(linalg.norm(b, 2))
```

```python
b = np.array([3, 4, 10])

# по умолчанию считается L2 норма
print(linalg.norm(b))
```

```python
# вычислим L2 норму для поставщиков
print(f'L2 for shop_1: {linalg.norm(shop_1, 2)}')
print(f'L2 for shop_2: {linalg.norm(shop_2, 2)}')
```

### Норма Чебышёва
максимум по модулю всех координат
$\\|\vec{a}|\\_\infty = \max_{i=1,\dots,n} |a_i|$ — $L_\infty$ норма;

```python
c = np.array([3, 4, -20])

# второй параметр - порядковый номер нормы
print(linalg.norm(c, np.inf))
```

# Нахождение угла межд векторами
$$\cos(\vec{a}, \vec{b}) = \frac{\vec{a} \cdot \vec{b}}{|\vec{a}| \cdot |\vec{b}|}.$$

```python
a = np.array([4, 8, 18])
b = np.array([-2, 3, 7])

cos_ab = a @ b / linalg.norm(a) / linalg.norm(b) # безразмерная величина

cos_ab_rad = np.arccos(cos_ab) # считаем в радианах
cos_ab_deg = np.rad2deg(cos_ab_rad) # переводим в градусы

print(f'Значение косинуса угла: {cos_ab}', f'Значение в радианах: {cos_ab_rad}', f'Значение в градусах: {cos_ab_deg}', sep='\n')
```

```python
k = np.array([12, 8, 3, 1])
i = np.array([0.7, 22, 5, 9])

i_k = i @ k

k_2 = linalg.norm(k)
i_2 = linalg.norm(i)

cos_ik = i_k / k_2 / i_2

cos_rad = np.arccos(cos_ik)
cos_deg = np.rad2deg(cos_rad)

print(cos_ik, cos_rad, cos_deg, sep='\n')
```

В пространстве расположен треугольник ABC. Координаты его вершин: A(7, 2, 11), B(5, 3, −4), C(−6, 8, 1).
Выведите на экран косинус ∠C 

```python
# задаем вектора от точки O=(0, 0, 0), которые идут до вершин треугольника
OA = np.array([7, 2, 11])
OB = np.array([5, 3, -4])
OC = np.array([-6, 8, 1])

# ищем координаты векторов, т.е. ребер треугольника: конец - начало
AC = OC - OA
BC = OC - OB

# ищем скалярное произведение AC и BC
a_c_sc = AC @ BC

# ищем L2 нормы
ac_2 = linalg.norm(AC)
bc_2 = linalg.norm(BC)

cos_angle_c = a_c_sc / (ac_2 * bc_2)
angle_rad = np.arccos(cos_angle_c)
angle_deg = np.rad2deg(angle_rad)

print(cos_angle_c, angle_rad, angle_deg, sep='\n')
```

# Расстояние между векторами
## через норму
- Расстояние норма разности двух векторов
- Норма может быть любой
- Расстояние между векторами для оценки похожести двух векторов вычисляют 
- Обычно это L2 норма. 

```python
# найдем 3 основных расстояния между векторами a и b
a = np.array([22, 4, -6, 3.9, 7])
b = np.array([12, -0.5, 6, 42, 1])

r_1 = linalg.norm(a-b, 1)
r_2 = linalg.norm(a-b, 2)
r_inf = linalg.norm(a-b, np.inf)

print(f'L1 норма: {r_1}\nL2 норма: {r_2}\nL_inf норма: {r_inf}')
```

```python
# неважно, что из чего вычитать - расстояние не изменится
linalg.norm(a-b, 1) == linalg.norm(b-a, 1), linalg.norm(a-b, 2) == linalg.norm(b-a, 2)
```

## через косинусное расстояние
- Так же для схожести векторов оценивают угол между ними
- Чем меньше угол - тем больше косинус => векторы более похожи
- Диапазон от -1 до 1

пример: есть 3 отзыва и нужно понять, какие из них похожи

1) Эти огурцы горькие, не понравились. Вердикт - не брать.
2) Огурцы отстой, не вкусные.
3) Супер, хрустят, рекомендую!

Выделим таблицу со всеми словами и посмотрим частоту слов в каждом отзыве. Получится 12 мерный вектор, который описывает отзыв

| эти |огрурцы | горькие | отстой | супер | не | понравились| вкусные | хрустят | вердикт | брать | рекомендую


```python
# посчитаем кол-во слов в трех отзывах и посмотрим, какие отзывы похожи
o_1 = np.array([1, 1, 1, 0, 0, 2, 1, 0, 0, 1, 1, 0])
o_2 = np.array([0, 1, 0, 1, 0, 1, 0, 1, 0, 0, 0, 0])
o_3 = np.array([0, 0, 0, 0, 1, 0, 0, 0, 1, 0, 0, 1])

r_1_2 = linalg.norm(o_1-o_2)
r_1_3 = linalg.norm(o_1-o_3)
r_2_3 = linalg.norm(o_3-o_2)

print("Отзывы 1 и 2:", r_1_2)
print("Отзывы 1 и 3:", r_1_3)
print("Отзывы 2 и 3:", r_2_3)
```

наименьшее расстояние между отзывами 2 и 3, но по своему содержанию отзывы вообще не похожи. Это связано с тем, что первый отзыв заметно длиннее других => у него больше нулевых значений. L2 норма учитывает размер векторов, поэтому разница отразилась в результатах. Нужно рассчитать косинусное расстояние

```python
cos_1_2 = o_1 @ o_2 / linalg.norm(o_1) / linalg.norm(o_2)
cos_1_3 = o_1 @ o_3 / linalg.norm(o_1) / linalg.norm(o_3)
cos_2_3 = o_2 @ o_3 / linalg.norm(o_2) / linalg.norm(o_3)
print("Отзывы 1 и 2:", cos_1_2)
print("Отзывы 1 и 3:", cos_1_3)
print("Отзывы 2 и 3:", cos_2_3)
```

Чем ближе косинус к 1, тем более сонаправлены и близки векторы. Косинусное расстояние не учитывает масштаб векторов (а соответственно, и длину отзыва), поэтому результаты получились адекватные: больше всего похожи друг на друга отзывы 1 и 2.

Итог: Разные метрики хороши для разных ситуаций. Евклидово расстояние учитывает размеры векторов, и его используют для анализа числовых характеристик.

Косинусное расстояние не зависит от масштаба векторов и не чувствительно к числовым характеристикам. Зато оно подходит для оценки сходства текстов.

На практике часто встречается ситуация, когда значения одного из признаков существенно отличаются от значений остальных. Влияние этого признака становится большим, хотя в реальности признаки должны быть равнозначимы. 

Чтобы сделать признаки равнозначимыми, их значения нормируют — приводят к единой шкале, например, к отрезку [0, 1]. Один из популярных способов такой: 

1.	Взять значения одного признака. Найти среди них наименьшее и наибольшее: Xmin и Xmax.
2.	Для каждого элемента этого признака вычислить нормированное значение по формуле $\frac{X - X_{\min}}{X_{\max} - X_{\min}}$
3.	Повторить шаги для остальных признаков.


```python
np.stack([m1, m2, m3, avg])
```

```python
# axis=0 - работа по столбцам
np.stack([m1, m2, m3, avg]).max(axis=0)
```

```python
# axis=1 - работа по строкам - из каждого вектора берем максимум
np.stack([m1, m2, m3, avg]).max(axis=1)
```

```python
np.stack([m1, m2, m3, avg]).min(axis=0)
```

```python
matrix = np.stack([m1, m2, m3, avg])
matrix
```

```python
# из первого столба вычитается 1, из второго 2
data = np.arange(1, 9)
matrix - data
```

```python
def get_norm_matrix_min_max(matrix: np.ndarray) -> np.ndarray:
    min_values = matrix.min(axis=0)
    max_values = matrix.max(axis=0)

    return (matrix - min_values) / (max_values - min_values)
```

```python
new_matrix = get_norm_matrix_min_max(np.stack([m1, m2, m3, avg]))
```

```python
m1_norm = new_matrix[0]
m2_norm = new_matrix[1]
m3_norm = new_matrix[2]
avg_norm = new_matrix[3]
```

```python
r1_norm = np.linalg.norm(m1_norm - avg_norm)
r2_norm = np.linalg.norm(m2_norm - avg_norm)
r3_norm = np.linalg.norm(m3_norm - avg_norm)

print(r1_norm, r2_norm, r3_norm, sep='\n')
```
